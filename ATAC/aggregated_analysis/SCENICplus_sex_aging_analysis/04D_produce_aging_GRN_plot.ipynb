#### Produce regulon plots for regulons that are increased in fetal state using ggplot

```r
library(tidyverse)
```

```r
aging_res_df = read.csv("04C_aging_sig_regulon_df.csv")

#fetal_res_df$shortened_regulon <- sub("_.*", "", fetal_res_df$regulon)
aging_res_df$log_padj <- -log10(aging_res_df$padj)
```

```r
# filter only to those cell-type specific regulons with mean AUC > 0.05
filtered_aging_res_df <- aging_res_df[aging_res_df$AUC_mean_young > 0.05, ]
```

```r
filtered_aging_res_df <- filtered_aging_res_df %>%
  mutate(direction = ifelse(log2FC < 0, "negative", "positive"))

# filter to by log2FC
filtered_aging_res_df <- filtered_aging_res_df[abs(filtered_aging_res_df$log2FC) > 0.25, ]

filtered_aging_res_df$regulon <- factor(
  filtered_aging_res_df$regulon,
  levels = sort(unique(filtered_aging_res_df$regulon))
)
```

```r
filtered_aging_res_df
```

```r
options(repr.plot.width = 5, repr.plot.height = 8)

p1 <- ggplot(data = filtered_aging_res_df, 
       mapping = aes(x = log2FC, y = regulon, size = log_padj, fill = log2FC)) +
  geom_point(shape = 21) + 
  facet_grid(~cell_type) + 
  scale_fill_gradient2(low = "blue", mid = "white", high = "red", midpoint = 0,
                        name = "log2FC") +
  theme_minimal() +              
  theme(
    axis.text.y = element_text(size = 16),
    axis.text.x = element_text(angle = 45, hjust = 1, size = 16),
    axis.title.y = element_text(size = 18), 
    axis.title.x = element_text(size = 18), 
    strip.text = element_text(face = "bold", size = 18),
    plot.title = element_text(size = 18),
    legend.position = "right") +
  labs(
    x = "log2 Fold Change",
    y = "Regulon",
    title = "Regulons with significant activity changes during aging (old vs. young)"
  )

p1

#ggsave(p1, filename = "04D_fetal_up_GRNs.pdf", height = 10, width = 15)
```
