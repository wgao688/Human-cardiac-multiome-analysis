### Subcluster annotation

- Cluster 0 are capillary (NR2F2/FLT4/LMO2)
- Cluster 1 are valvular (NRP1/FOXC2/VEGFC)
- Cluster 2 are fetal-enriched (CELSR1/WNT7A/WNT11)
- Cluster 3 are activated/inflammatory (GSEA is enriched for TNF-alpha signaling/Myc)
- Cluster 4 is a very small cluster from one study; merge with major cluster 0

```python
adata.obs['subcluster'] = adata.obs[leiden_key].map(
    {"0": "Capillary",
     "1": "Valvular",
     "2": "Fetal-enriched",
     "3": "Activated",
     "4": "Capillary",
    }
)

subcluster_order = ["Activated", "Capillary", 
                    "Fetal-enriched", "Valvular",]

adata.obs['subcluster'] = pd.Categorical(
    adata.obs['subcluster'], categories=subcluster_order, ordered=True
)

subcluster_colors = {
    "Valvular": "lightblue",   
    "Capillary": "orange",  
    "Fetal-enriched": "pink", 
    "Activated": "yellow"
}

# apply colors to scanpy
adata.uns['subcluster_colors'] = [subcluster_colors[sc] for sc in subcluster_order]
```

### Produce plots

```python
with plt.rc_context(): 
    sc.pl.umap(adata, color = ['subcluster'], legend_loc = "on data", palette = subcluster_colors, show=False)
    plt.savefig(plots_dir + cell_type + "_subclustered.pdf", bbox_inches="tight")
```
