#### We run into an issue with too many cells, see here https://github.com/satijalab/seurat/issues/4030 if we don't split the adata, therefore we will split the adata here, and then combine later

#### Since NEBULA should be run at a cell-type level, divide the adata into individual cell type adatas

We will perform NEBULA analysis using only the non-diseased donors to look for changes across sex and aging.

```python
import scanpy as sc 
import os 
from collections import Counter
```

```python
cell_type_adata_dir = "cell_type_adata/"
os.makedirs(cell_type_adata_dir, exist_ok=True)
```

```python
%%time
adata = sc.read_h5ad("../07_final_RNA_without_scvi.h5ad")
adata
```

```python
cell_types = adata.obs.final_cell_type.unique()
len(cell_types)
```

### Filter to only the non-diseased postnatal donors

```python
filt_adata = adata[(adata.obs.disease_binary == "N") & (adata.obs.age_status != "fetal"), :].copy()
```

```python
filt_adata.shape
```

### Subset to each cell type

```python
%%time
# subset to each cell type
for cell_type_val in cell_types:
    print(cell_type_val, flush=True)
    subset_adata = filt_adata[filt_adata.obs.final_cell_type == cell_type_val].copy()
    subset_adata.write(cell_type_adata_dir + cell_type_val + ".h5ad")
```
