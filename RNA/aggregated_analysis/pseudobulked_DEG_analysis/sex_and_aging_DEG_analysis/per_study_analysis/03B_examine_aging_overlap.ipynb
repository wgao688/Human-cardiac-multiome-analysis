```python
import pandas as pd
import numpy as np
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
def produce_heatmap(df, figsize=(8, 10)):
    heatmap_df = df.set_index("id").drop(columns=["col_sums"])

    plt.figure(figsize=figsize)
    ax = sns.heatmap(
        heatmap_df,
        cmap="Reds",        # 0 = white, 1 = red
        linewidths=0.5,
        vmin=0, vmax=1,     # force 0–1 range
        cbar=False          # hide colorbar
    )

    plt.title("Shared DEGs Across Studies", fontsize=16, pad=20)
    plt.xlabel("Study", fontsize=14)
    plt.ylabel("Gene", fontsize=14)
    plt.tight_layout()
    return plt
```

### CM analysis

```python
CM_old_up = pd.read_csv("age_results_dir/Cardiomyocyte_up_overlap_results.csv", index_col = 0)
CM_old_up.sort_values(by = "col_sums", ascending=False).head(13)

CM_old_up_in_at_least_two = CM_old_up[CM_old_up['col_sums'] > 2]
produce_heatmap(CM_old_up_in_at_least_two, figsize=(4, 7))
```

```python
CM_young_up = pd.read_csv("age_results_dir/Cardiomyocyte_down_overlap_results.csv", index_col = 0)
CM_young_up.sort_values(by = "col_sums", ascending=False).head(13)

CM_young_up_in_at_least_two = CM_young_up[CM_young_up['col_sums'] > 2]
produce_heatmap(CM_young_up_in_at_least_two, figsize=(4, 7))
```

```python
Fibro_young_up = pd.read_csv("age_results_dir/Fibroblast_down_overlap_results.csv", index_col = 0)
Fibro_young_up.sort_values(by = "col_sums", ascending=False).head(15)

Fibro_young_up_in_at_least_two = Fibro_young_up[Fibro_young_up['col_sums'] > 1]
produce_heatmap(Fibro_young_up_in_at_least_two, figsize=(4, 7))
```

```python
Fibro_old_up = pd.read_csv("age_results_dir/Fibroblast_up_overlap_results.csv", index_col = 0)
Fibro_old_up.sort_values(by = "col_sums", ascending=False).head(15)

Fibro_old_up_in_at_least_two = Fibro_old_up[Fibro_old_up['col_sums'] > 1]
produce_heatmap(Fibro_old_up_in_at_least_two, figsize=(4, 7))
```
