### Produce concordant barplot

```r
library(tidyverse)
```

```r
options(repr.plot.width = 5, repr.plot.height = 4)

DCM_df <- data.frame(
  category = c("Fully concordant", "Partially concordant", "Discordant"),
  count = c(16, 48, 15)
)

DCM_df$category <- factor(DCM_df$category, c("Fully concordant", "Partially concordant", "Discordant"))

p1 <- ggplot(DCM_df, aes(x = category, y = count, fill = category)) +
  geom_bar(stat = "identity", width = 0.9) +  # wider bars = less space between
  scale_fill_manual(values = c(
    "Fully concordant" = "#4CAF50",    # green
    "Partially concordant" = "#FFC107",# yellow-orange
    "Discordant" = "#F44336"           # red
  )) +
  geom_text(aes(label = count), vjust = -0.4, size = 5) +
  theme_bw(base_size = 14) +
  labs(title = "Concordance summary", x = "", y = "# GWAS loci") +
  theme(
    legend.position = "none",
    axis.text.x = element_text()
  )

p1

ggsave(p1, filename = "02C_concordance_DCM_barplot.pdf", width = 5, height = 4)
```
