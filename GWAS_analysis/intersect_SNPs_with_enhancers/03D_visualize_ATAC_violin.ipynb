#### DTL peak corresponds to chr1:211932524-211933025

```r
DTL_peak_df <- read.csv("03C_DTL_ATAC_df.csv", row.names = 1)
DTL_peak_df$age_disease_status <- factor(DTL_peak_df$age_disease_status, c("postnatal:N", "postnatal:Y", "fetal:N"))
```

```r
options(repr.plot.width = 8, repr.plot.height = 6)

# produce plots
p1 <- ggplot(data = DTL_peak_df, aes(x = age_disease_status, y = expression)) +
  geom_violin(trim = FALSE, fill = "grey85", color = "black") +
  geom_boxplot(width = 0.2, outlier.shape = NA, fill = "grey85", color = "black") +
  geom_jitter(width = 0.15, size = 1.5, alpha = 0.6) +
  theme_bw() +
  labs(x = "Age/Disease Status", y = "VST-normalized counts", title = "DTL linked peak accessibility") +
  theme(legend.text = element_text(size = 16), 
         legend.title = element_text(size = 16, color = "black"), 
         plot.title = element_text(size = 24, hjust = 0.5, color = "black"), 
         axis.title.x = element_text(size = 16, color = "black"),
         axis.title.y = element_text(size = 16, color = "black"),
         axis.text.x = element_text(size = 16, color = "black"),
         axis.text.y = element_text(size = 16, color = "black"),
	 strip.text = element_text(size = 18, color = "black")) + ylim(0, 8)

p1
```

#### also plot version without diseased, since this is acute MI, not heart failure

```r
options(repr.plot.width = 6, repr.plot.height = 5)

filt_DTL_peak_df <- DTL_peak_df[DTL_peak_df$age_disease_status != "postnatal:Y", ]

# produce plots
p1_filt <- ggplot(data = filt_DTL_peak_df, aes(x = age_disease_status, y = expression)) +
  geom_violin(trim = FALSE, fill = "grey85", color = "black") +
  geom_boxplot(width = 0.2, outlier.shape = NA, fill = "white", color = "black") +
  geom_jitter(width = 0.15, size = 1.5, alpha = 0.6) +
  theme_bw() +
  labs(x = "Age/Disease Status", y = "VST-normalized counts", title = "DTL linked peak accessibility") +
  theme(legend.text = element_text(size = 16), 
         legend.title = element_text(size = 16, color = "black"), 
         plot.title = element_text(size = 24, hjust = 0.5, color = "black"), 
         axis.title.x = element_text(size = 16, color = "black"),
         axis.title.y = element_text(size = 16, color = "black"),
         axis.text.x = element_text(size = 16, color = "black"),
         axis.text.y = element_text(size = 16, color = "black"),
	 strip.text = element_text(size = 18, color = "black"))

ggsave(p1_filt, filename = "03D_DTL_ATAC_peak.pdf", width = 8, height = 6)
```
