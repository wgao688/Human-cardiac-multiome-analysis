### Identify fetalization genes that overlap with the DCM/HCM prioritized genes 

Use the genes prioritized using our method and the original papers

```python
DCM_df = pd.read_csv("02B_prioritized_DCM_genes.csv", index_col = 0)
HCM_df = pd.read_csv("02B_prioritized_HCM_genes.csv", index_col = 0)
```

```python
fetalization_gene_dir = "../../../../RNA/aggregated_analysis/v3_pseudobulked_analysis/intersection_analysis/fetalization_genes/"
```

```python
cell_types = ["Cardiomyocyte", "Endocardial", "Endothelial", "Fibroblast", "LEC", "Lymphoid", 
              "Myeloid", "Neuronal", "Pericyte", "vSMC"]

fetalization_gene_list = list()

for cell_type in cell_types:
    print(cell_type)
    fetalization_CM_genes = pd.read_csv(fetalization_gene_dir + cell_type + "_up_fetalization_genes.csv")
    fetalization_CM_genes.columns = ['gene']
    fetalization_CM_genes['cell_type'] = cell_type
    fetalization_gene_list.append(fetalization_CM_genes)

fetalization_all_df = pd.concat(fetalization_gene_list, ignore_index=True)
```

```python
concordant_DCM_df = DCM_df[DCM_df['concordant'] == True]
concordant_HCM_df = HCM_df[HCM_df['concordant'] == True]
```
