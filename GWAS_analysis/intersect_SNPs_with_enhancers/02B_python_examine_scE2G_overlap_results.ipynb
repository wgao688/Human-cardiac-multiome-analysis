```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
cell_types = ["Fibroblast", "Endothelial", "Epicardial", "Lymphoid", "Myeloid", "Neuronal", "Pericyte", "vSMC"]
```

```python
res_list = list()

for cell_type in cell_types:
    res_df = pd.read_csv("intersecting_SNPs/" + cell_type + "_loci_intersections.csv")
    filt_res_df = res_df[["TargetGene", "lead_SNP", "disease", 
                          "candidate_SNP", "RNA_pseudobulkTPM", "CellType", "prioritized_gene"]]
    res_list.append(filt_res_df)
```

```python
concat_df = pd.concat(res_list)
```
