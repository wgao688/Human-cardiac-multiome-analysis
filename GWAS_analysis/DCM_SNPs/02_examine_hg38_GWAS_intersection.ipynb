### Examine intersection between the loci for Jurgens and Zheng in hg38

```python
import pandas as pd
import numpy as np
```

```python
Jurgens_df = pd.read_csv("Jurgens_GWAS_hits_GRCh38.bed", delim_whitespace=True, header=None)
Jurgens_df.columns = ["chr", "start", "end", "rsID", "prioritized_gene", "total_score"]
Jurgens_df.head()
```

```python
Zheng_df = pd.read_csv("Zheng_GWAS_hits_GRCh38.bed", delim_whitespace=True, header=None)
Zheng_df.columns = ["chr", "start", "end", "rsID", "prioritized_gene", "total_score"]
Zheng_df.head()
```

```python
intersecting_SNPs = list( set(Zheng_df.rsID) & set(Jurgens_df.rsID) )
print(f"Number of intersecting SNPs: {len(intersecting_SNPs)}")
intersecting_SNPs
```

```python
Jurgens_df[Jurgens_df['rsID'].isin(intersecting_SNPs)]
```

```python
Zheng_df[Zheng_df['rsID'].isin(intersecting_SNPs)]
```
