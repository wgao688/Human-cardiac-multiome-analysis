### Combine all multiome ATAC and all multiome RNA, separately

```python
import scanpy as sc
```

```python
%%time
Kanemaru_ATAC_adata = sc.read_h5ad("01_filtered_Kanemaru_multiome_ATAC_adata.h5ad")
ENCODE_ATAC_adata = sc.read_h5ad("01_filtered_ENCODE_multiome_ATAC_adata.h5ad")
```

```python
%%time
Kanemaru_RNA_adata = sc.read_h5ad("01_filtered_Kanemaru_multiome_RNA_adata.h5ad")
ENCODE_RNA_adata = sc.read_h5ad("01_filtered_ENCODE_multiome_RNA_adata.h5ad")
```

```python
%%time
combined_ATAC_adata = sc.concat([Kanemaru_ATAC_adata, ENCODE_ATAC_adata])
```

#### Need to add back reference sequences

```python
combined_ATAC_adata.uns['reference_sequences'] = Kanemaru_ATAC_adata.uns['reference_sequences']
```

```python
combined_ATAC_adata.write("02_combined_multiome_ATAC.h5ad")
```

```python
%%time
combined_RNA_adata = sc.concat([Kanemaru_RNA_adata, ENCODE_RNA_adata])
combined_RNA_adata
```

```python
combined_RNA_adata.write("02_combined_multiome_RNA.h5ad")
```
