### Create bed file for bedtools analysis and .txt list of SNPs for PLINK

```python
import numpy as np
import pandas as pd
```

```python
Tadros_GWAS_loci = pd.read_csv("Tadros_GWAS_hits_GRCh38.bed", delim_whitespace=True, header=None)
Tadros_GWAS_loci.columns = ["chr", "start", "end", "rsID", "candidate_gene"]
Tadros_GWAS_loci['study'] = "Tadros"
Tadros_GWAS_loci.head()
print(f"Number of SNPs: {Tadros_GWAS_loci.shape[0]}")
```

### Save as bed file

```python
GWAS_loci_bed = Tadros_GWAS_loci[["chr", "start", "end"]]
GWAS_loci_bed.to_csv("GWAS_loci.bed", header=False, index=None, sep = "\t")
```

### For plink, we will want the format of `<chr>:start-end` without any `"chr"`

```python
GWAS_loci_bed['plink_name'] = GWAS_loci_bed['chr'].str.replace("chr","") + ":" + GWAS_loci_bed['start'].astype(str) 
GWAS_loci_bed['plink_name'].to_csv("hg38_combined_unique_HCM_SNPs.csv", header=False, index=False)
```
